# Getting started with pyDARTLAB

pyDARTLAB is a Python version of [DART_LAB](https://github.com/NCAR/DART/tree/main/guide/DART_LAB),
the interactive ensemble data assimilation tutorial distributed with NCAR's
**DART** (Data Assimilation Research Testbed). The tutorial is a series of
notebooks, each pairing short theory sections with hands-on interactive tools:

| Notebook | Topic |
|---|---|
| 01 | Ensemble data assimilation concepts in 1D |
| 02 | Multivariate assimilation: how observations impact unobserved variables |
| 03 | Inflation and localization |
| 04 | Non-Gaussian and bounded filters (QCEFF) |
| 05 | Adaptive inflation |
| 06 | Using the real DART system |

## How the interactive tools work

Every notebook starts with `%matplotlib widget`, which turns matplotlib
figures into live widgets (via [ipympl](https://matplotlib.org/ipympl/)).
Each DART_LAB tool is launched with one line, e.g. `apps.oned_ensemble()`,
and renders a figure with controls beside it.

Several tools let you **create an ensemble by clicking in the plot** —
click to place members, then click *outside* the axes to finish, exactly
like the MATLAB tools.

> **No mouse events?** (e.g. on Colab or a static viewer): every
> click-driven app also has a `set_ensemble` method, so all exercises can
> be done programmatically. Each exercise below shows both routes.

````python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
import pydartlab as dl
import pydartlab.apps as apps
````

## Color conventions

All DART_LAB materials use the same color scheme:

* <span style="color:#008000">**green = prior**</span> (the forecast, before assimilation)
* <span style="color:#D70A53">**red = observation / likelihood**</span>
* <span style="color:#0000FF">**blue = posterior**</span> (the analysis, after assimilation)

The palette is settable. For a colorblind-friendly (Okabe-Ito) palette run
the cell below; `dl.style.set_colors(prior=...)` changes individual roles, and
`dl.style.use_dartlab_palette()` restores the defaults.

````python
# Uncomment to switch palettes:
# dl.style.use_colorblind_palette()
# dl.style.set_colors(prior="#009E73")
dl.style.colors
````

## A first tool

The cell below opens the simplest tool, `gaussian_product` (used in

and check that the figure updates. If it does, your environment is ready

The app object is returned, so its state can also be driven from code —

````python
app = apps.gaussian_product()
app
````

````python
# The same thing, mouse-free:
app.prior_mean.value, app.prior_sd.value = 0.0, 2.0
app.obs_mean.value, app.obs_sd.value = 2.0, 1.0
app.update()
print("posterior (mean, sd, weight):", app.last_result)
````

## The package layers

Everything the apps do is available as plain functions and classes:

* `pydartlab` — filter algorithms (`obs_increment_eakf/enkf/rhf`...),
  inflation, localization, the Lorenz models, ...
* `pydartlab.experiments` — scriptable cycling experiments
  (`OneDExperiment`, `Lorenz96Experiment`, ...) used by the cycling apps.
  Use these to run hundreds of assimilation cycles in a loop and plot the
  results — something the notebooks do alongside the interactive tools.
